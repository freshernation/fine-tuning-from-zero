## 5. Your loss curve

This only works once the milestone passes.

```python
import importlib.util
from pathlib import Path

import matplotlib.pyplot as plt

spec = importlib.util.spec_from_file_location("train_bigram", "week-01/milestone/train_bigram.py")
train_bigram = importlib.util.module_from_spec(spec)
spec.loader.exec_module(train_bigram)

result = train_bigram.run(Path("data/tiny.txt").read_text())

steps = [row["step"] for row in result["history"]]
plt.plot(steps, [row["train"] for row in result["history"]], label="train")
plt.plot(steps, [row["val"] for row in result["history"]], label="val")
plt.xlabel("step")
plt.ylabel("loss")
plt.legend()
plt.show()

print(result["sample"])
```
